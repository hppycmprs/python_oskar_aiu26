- arithmetic sum

- 1+2+3+...+9+10
- 1+2+3+...+99+100

```python
sum_ = 0

for i in range(1, 11):
    sum_ += i

    if i == 10:
        print(f"{i}", end="=")
    else:
        print(f"{i}", end="+")

print(sum_)
```

# random
- count number of fives in 10_000_000 dice rolls -> calculate proportion of fives

```python
import random as rnd 

# type hinting
number_rolls: int = 10_000_000
count_fives: int = 0

for roll in range(number_rolls):
    dice = rnd.randint(1,6)

    if dice == 5:
        count_fives += 1

count_fives/number_rolls
```

# nested loops

```python
for row in range(3):
    for col in range(3):
        print(f"({row},{col})", end=" ")
    print()
```
