## Try except

- try-block -> test dangerous code
- except-block -> handles error instead of crashing
- raise -> throw exception

```python
try:
    age = int(input("What is your age? "))

    if not 0 <= age < 125:
        raise ValueError(f"Age must be between 0-124, not {age}")
    print(f"You are {age} years old")
except ValueError as err:
    print(err)
```

```python
while True:
    age = input("What is your age? ")              # eller round(float(input("What is your age?"))) round av float av input

    try:
        if not age.isdigit():
            raise TypeError(f"Age must be a number, not {type(age)}")

        age = float(age)

        if not 0 <= age < 125:
            raise ValueError(f"Age must be between 0-124, not {age}")
        print(f"You are {age:.0f} years old")
        break
    except (TypeError, ValueError) as err:
        print(err)
```
