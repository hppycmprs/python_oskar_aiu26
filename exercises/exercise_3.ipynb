# 0. Dice rolls 

```python
# Simulate 10 dice rolls and append the rolls to a list. 
import random as rnd

number_rolls = 10
dice_roll_list = []

for roll in range(number_rolls):
    dice_rolls = rnd.randint(1,6)
    dice_roll_list.append(dice_rolls) # .append adds dice_rolls to list

print(dice_roll_list)
```

```python
# Do this once again but now use list comprehension instead to create the list.

dice_roll_list_2 = [rnd.randint(1,6) for roll in range(number_rolls)]
dice_roll_list_2.sort() # a) sorted in ascending order
print(dice_roll_list_2)
```

```python
dice_roll_list_2 = [rnd.randint(1,6) for roll in range(number_rolls)]
dice_roll_list_2.sort(reverse=True) # b) sorted in descending order
print(dice_roll_list_2)
```

```python
# c) find the maximum and minimum value in the list

dice_roll_list_2 = [rnd.randint(1,6) for roll in range(number_rolls)]
dice_roll_list_2.sort()

print(dice_roll_list_2)
print(dice_roll_list_2[0]) # prints the minimum value in list, in this case the first roll since its in ascending order
print(dice_roll_list_2[9]) # prints the maximum value in list, -||-
```

```python
# d) calculate the mean (average) of the list

dice_roll_list_2 = [rnd.randint(1,6) for roll in range(number_rolls)]

print(dice_roll_list_2)
average = sum(dice_roll_list_2) / len(dice_roll_list_2) # sum of roll numbers in list / length of number of elements in list saved in variable "average"
print(average)
```

```python
# e) find the median of the list

dice_roll_list_2 = [rnd.randint(1,6) for roll in range(number_rolls)]
dice_roll_list_2.sort()

print(dice_roll_list_2)
median = (dice_roll_list_2[4] + dice_roll_list_2[5]) / 2 # because list contains even amount of index numbers, take the 2 middle elements and divide them by 2 to get median
print(median)
```

```python
# f) extract only the unique values from the list. The output should be a list

dice_roll_list_2 = [rnd.randint(1,6) for roll in range(number_rolls)]
dice_roll_list_2.sort()

print(dice_roll_list_2)

set_of_unique = set(dice_roll_list_2) # making set of list to exclude duplicates and naming it set_of_unique
dice_roll_list_unique = list(set_of_unique) # making a new list called dice_roll_list_unique out of the set

print(dice_roll_list_unique)
```

# 1. Food menu

```python
# Create a:

# a) list with the following elements: "vegetarian lasagna", "spaghetti", "fish", "soup", "pancakes". (*)

# b) list with the weekdays (*)

# c) create a food menu with each day corresponding to each food item and print it out like this (*)

# Day        Food
# -----------------
# Monday     vegetarian lasagna
# Tuesday    spaghetti
# Wednesday  fish
# Thursday   soup
# Friday     pancakes

foods = ["vegetarian lasagna", "spaghetti", "fish", "soup", "pancakes"]
weekdays = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday"]

menu = zip(weekdays, foods)
print("Day        Food")
print("-----------------")
for weekday, food in menu:
    print(f"{weekday:10} {food}")
```

# 4. Dice rolls convergence (**)

```python
# Simulate:

# a) 100 dice rolls and count the number of outcome six. (*)

import random as rnd

number_throws = 100
sixes = 0

for throw in range(number_throws):
    value = rnd.randint(1,6)
    if value == 6:
        sixes += 1

print(sixes)
```

```python
# b) 10, 100, 1000, 10000, 100000, 1000000 dice rolls. Count the number of outcome six in each simulation and store it in a list. 
# Compute the probability of outcome six in each simulation. (**)

throw_ranges = [10, 100, 1000, 10000, 100000, 1000000]
sixes_list = []
probability_list = []

for throw_range in throw_ranges:
    six_prevalence = 0
    for throw in range(throw_range):
        the_value = rnd.randint(1,6)
        if the_value == 6:
            six_prevalence += 1
    sixes_list.append(six_prevalence)
    probability = six_prevalence/throw_range
    probability_list.append(probability)

print(sixes_list)
print(probability_list)
```
