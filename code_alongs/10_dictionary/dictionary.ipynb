## Dictionary

```python
person = dict(
    name="Kokchun",
    age=35,
    work="teacher",
    interests=["math", "python", "linux"],
    phone=112,
) # här är name, age, work, interests, phone = keys

person
```

```python
person = {
 'name': 'Kokchun',
 'age': 35,
 'work': 'teacher',
 'interests': ['math', 'python', 'linux'],
 'phone': 112
 } # samma sak som ovan fast med curly braces {}

person
```

```python
print(f"{person["name"] = }")
print(f"{person["age"] = }")
print(f"{person[interests]}")
```

## Looping through a dict

```python
for key in person:
    print(f"{key}: {person[key]}")
```

```python
for key, value in person.items():  # tuple unpacking
    print(f"{key}: {value}")
```

```python
person["has_driving_license"]=True
person
```

```python
"ABCDEF"[::-1] # reverse, går baklänges, slicing fast baklänges
```

```python
# F<10, E:10, D:20, C:30, B:40, A:50

#empty dict
grade_limits= {}

for i, char_ in enumerate("ABCDEF"[::-1]): # eller skriv FEDCBA från början
    grade_limits[char_] = i*10 # assigns new keys and values to dict

grade_limits["F"] = "<10"
grade_limits
```

# Betygsgränser
# A: 50
# B: 40
# ...

```python
for grade, limit in reversed(grade_limits.items()):
    print(f"{grade}: {limit}")
```

```python
grades = sorted(grade_limits.keys())

for grade in grades:
    print(f"{grade}: {grade_limits[grade]}")
```

# Dictionary comprehension

```python
grade_limits = {grade: i*10 for i, grade in enumerate("FEDCBA")} # grade = nyckel   grade: i*10 = key: värde
grade_limits
```
