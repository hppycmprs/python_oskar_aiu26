# 2. Medicine

```python
#Adults & adolescent over 40 kg  	over 12 years   	1-2
#Children 26-40 kg               	7-12 years	        1/2-1
#Children 15-25 kg	                3-7 years	        1/2

ad_weight = 41
ad_age = 13

child_upper_min_weight = 26
child_upper_max_weight = 40

child_upper_min_age = 7
child_upper_max_age = 12

child_lower_min_weight = 15
child_lower_max_weight = 25

child_lower_min_age = 3
child_lower_max_age = 7

user_age = int(input("Enter your age"))
user_kg = int(input("Enter your weight in kg"))

if user_age > 18 and user_kg < 26:
    print("Your weight is really low for your age, go see a doctor")

if user_age >= ad_age and user_kg >= ad_weight:
    print("Your recommended dose is 1-2 pills")
elif child_upper_min_weight <= user_kg <= child_upper_max_weight:
    print("Your recommended dose is 1/2-1 pill")
elif child_lower_min_weight <= user_kg <= child_lower_max_weight:
    print("Your recommended dose is 1/2 pill")
else:
    print("You weight is too low to ingest this pill")
```

# 6. Multiplication game 

```python
import random as rnd

score = 0

while True:
    x = rnd.randint(1, 10)
    y = rnd.randint(1, 10)
    correct_answer = x * y

    user_solution = int(input(f"What is {x}*{y}?"))

    if user_solution == correct_answer:
        print("Good work")
        score += 1
        user_choice = input(f"Your score is now {score} points.\nHit enter to play again or any other button to exit")
        if user_choice == "":
            continue
        else:
            print("Thanks for playing")
            break

    else:
        print(f"No, the correct answer is {correct_answer}")
        user_choice = input(f"Your score is now {score} points.\nHit enter to play again or any other button to exit")
        if user_choice == "":
            continue
        else:
            print("Thanks for playing")
            break
```
