# SOLID principles

```python
class Employee:
    def __init__(self, name, phone, salary):
        self.name = name
        self.phone = phone
        self.salary = salary

    def net_salary(self, tax_rate = 0.3):
        return self.salary * (1-tax_rate)

    def generate_report(self):
        with open(f"{self.name}.txt", "w") as file:
                  file.write("Employee report\n")
                  file.write(f"Name: {self.name}\nPhone: {self.phone}\nSalary: {self.salary}\nNet salary: {self.net_salary}")

e1 = Employee("Christer", "031-892121", 35000)
e1.generate_report()
```
