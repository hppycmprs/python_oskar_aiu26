```python
name = input("Enter your name: ")
print(f"Hello {name}")
```

```python

gravity = 9.82
mass = int(input("Enter your weight in kg: "))

force = mass*gravity

print (f"Your force is {force}N")
```

```python
import math

PI = math.pi

radius = 1
perimeter = 2*PI*radius
# **2 means to the power of 2
area = PI*radius**2

print(f"radius is {radius}, perimeter is {perimeter}, area is {area}")
```
