## Clean up quotes

```python
import re

with open("data/quotes.txt") as f_read, open("data/quotes_clean.txt", "w") as f_write:
    f_write.write("Cool quotes\n")

    for i, line in enumerate(f_read):
        line = line.strip(" \n")

        line = re.sub(" +", " ", line)  # substitute 1 spaces or more with 1 space

        f_write.write(f"{line}. {line}\n")

        print(line)
```
