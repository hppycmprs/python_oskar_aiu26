## Encapsulation, "private", property

Encapsulation
- hide implementation details
- property -> getter and setter

Private
- don't access this attribute/method from outside
- private by convention in Python _ or __ as prefix

## Private

```python
from datetime import datetime

class Student:
    def __init__(self, name, program, personal_number):
        self.name = name
        self.program = program
        self._personal_number = personal_number # private by convention
    
    def age(self):
        current_year = datetime.now().year
        birth_year = int("20"+self._personal_number[:2])

        age = current_year - birth_year
        return age

    def __str__(self):
        return f"Student {self.name} is {self.age()} years old and studies {self.program}"

    def __repr__(self):
        return f"Student('{self.name}', '{self.program}')"

s1 = Student("Babba", "AI26", "240303-9988")
s1
```

## An example: OldCoinStash

- conventions for classes Python: PascalCase
- conventions for variables: snake_case
- camelCase in javascript

```python
class OldCoinStash:
    def __init__(self, owner):
        self.owner = owner

        self._riksdaler = 0
        self._skilling = 0

    def deposit(self, riksdaler = 0, skilling = 0):
        if riksdaler < 0 or skilling < 0:
            raise ValueError(f"Give positive riksdaler and positive skilling, not {riksdaler} riksdaler and not {skilling} skilling")
        # more error handling
        self._riksdaler += riksdaler
        self._skilling += skilling
        

    def withdraw(self, riksdaler = 0, skilling = 0):
        if riksdaler > self._riksdaler or skilling > self._skilling:
            raise ValueError(f"Can't withdraw more than you have")
        self._riksdaler -= riksdaler
        self._skilling -= skilling
        

    def check_balance(self):
        print(f"Coins in stash: {self._riksdaler} riksdaler and {self._skilling} skillingar")

    def __repr__(self):
        return f"OldCoinStash(owner='{self.owner}')"

stash1 = OldCoinStash("Ragnar")
stash1.check_balance()

try:
    stash1.deposit(riksdaler=5)
except ValueError as err:
    print(err)

stash1.deposit(3,6)
stash1.check_balance()

stash1.withdraw(2,2)

stash1
```
