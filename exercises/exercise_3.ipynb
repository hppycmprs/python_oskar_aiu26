# 0. Dice rolls 

```python
# Simulate 10 dice rolls and append the rolls to a list. 
import random as rnd

number_rolls = 10
dice_roll_list = []

for roll in range(number_rolls):
    dice_rolls = rnd.randint(1,6)
    dice_roll_list.append(dice_rolls) # .append lägger till innehåll i en lista

print(dice_roll_list)
```

```python
# Do this once again but now use list comprehension instead to create the list.

dice_roll_list_2 = [rnd.randint(1,6) for roll in range(number_rolls)]
dice_roll_list_2.sort() # a) sorted in ascending order
print(dice_roll_list_2)
```

```python
dice_roll_list_2 = [rnd.randint(1,6) for roll in range(number_rolls)]
dice_roll_list_2.sort(reverse=True) # b) sorted in descending order
print(dice_roll_list_2)
```

```python
# c) find the maximum and minimum value in the list

dice_roll_list_2 = [rnd.randint(1,6) for roll in range(number_rolls)]
dice_roll_list_2.sort()

print(dice_roll_list_2)
print(dice_roll_list_2[0]) # prints the minimum value in list, in this case the first roll since its in ascending order
print(dice_roll_list_2[9]) # prints the maximum value in list, -||-
```

```python
# d) calculate the mean (average) of the list

dice_roll_list_2 = [rnd.randint(1,6) for roll in range(number_rolls)]

print(dice_roll_list_2)
average = sum(dice_roll_list_2) / len(dice_roll_list_2) # sum of roll numbers in list / length of number of elements in list saved in variable "average"
print(average)
```

```python
# e) find the median of the list

dice_roll_list_2 = [rnd.randint(1,6) for roll in range(number_rolls)]
dice_roll_list_2.sort()

print(dice_roll_list_2)
median = (dice_roll_list_2[4] + dice_roll_list_2[5]) / 2 # because list contains even amount of index numbers, take the 2 middle elements and divide them by 2 to get median
print(median)
```

```python
# f) extract only the unique values from the list. The output should be a list

dice_roll_list_2 = [rnd.randint(1,6) for roll in range(number_rolls)]
dice_roll_list_2.sort()

print(dice_roll_list_2)

set_of_unique = set(dice_roll_list_2) # making set of list to exclude duplicates and naming it set_of_unique
dice_roll_list_unique = list(set_of_unique) # making a new list called dice_roll_list_unique out of the set

print(dice_roll_list_unique)

```
