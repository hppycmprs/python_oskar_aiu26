## Oil leakage

There is an oil leakage causing the bird population of an island to be reduced by 1/2 each year. From start there were 80 000 birds, how many years will it take for the population to be reduced to 1/10?

```python
birds = 80_000
year = 0

while birds > 8000:
    print(f"year {year} we have {birds:.0f} birds")
    year += 1
    birds /= 2
```
