# OOP fundamentals

```python
class Person:
    def __init__(self, name, age, hair): ## dunder init -> will run when an instance is created
        print("__init__ is running")
        self.name = name
        self.age = age
        self.hair = hair

# instantiate an instance of the Person class and we give it values on the attributes
p1 = Person("Bob", 22, True)
print(f"{type(p1) = }")

# used dot operator to get the values of the attributes
print(f"{p1.name = }")
print(f"{p1.age = }")
print(f"{p1.hair = }")
```

```python
p1.name
```

```python
p2 = Person(age=25, hair=False, name="Harry")

p2.name, p2.age, p2.hair
```

## \_\_repr\_\_

"dunder repper"

- represents the object
- jupyter notebook prints this out if you put it in the end

```python
class Person:
    def __init__(self, name, age, hair):
        self.name = name
        self.age = age
        self.hair = hair

    def __repr__(self):
        return f"Person({self.name}, {self.age}, {self.hair})"

p3 = Person("Anna", 9, True)
p3
```

```python
p3.age = 19
p3
```

```python
p3.language = ["python, svenska"]
p3.language
```

```python
print(p3)
```

```python
print(p1)
```

## Type hinting and documentation

```python
class Person:
    """"
    Person class for representing a physical person

    Attributes: 
    ...

    Methods:
    ...

    Example usage:
    ...

    """
    def __init__(self, name: str, age: float, hair: bool = True):
        self.name = name
        self.age = age
        self.hair = hair

    def __repr__(self):
        return f"Person({self.name}, {self.age}, {self.hair})"
    
p4 = Person("Cedrik", 30)
p5 = Person("Cedrik", "trettio") # still okay
p4
```
