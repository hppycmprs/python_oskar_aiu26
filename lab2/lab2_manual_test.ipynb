### Add several circles and rectangles areas together, result should be 679,6371

```python
c1 = Circle(5, 5, 5)
c2 = Circle(6)
c3 = Circle(7)
c4 = Circle(8)

r1 = Rectangle(5, 5, 5, 5)
r2 = Rectangle(5, 6)
r3 = Rectangle(6, 6)
r4 = Rectangle(6, 7)

circles_and_rectangles = [c1, c2, c3, c4, r1, r2, r3, r4]

circles_and_rectangles # validating output before for loop
```

```python
area_sum = 0
for shape in circles_and_rectangles:
    area_sum += shape.area
print(area_sum)
```

### Testing the circle class

#### Testing valid circle

```python
c1 = Circle(5, 5, 5)
c2 = Circle(6)
c3 = Circle(7)
c4 = Circle(8)
c5 = Circle(1)

print(c1.radius, c1.x, c1.y)
print(c1.area, c1.perimeter, c1.unit_circle_or_not())
print(c2.radius, c2.x, c2.y)
print(c2.area, c2.perimeter, c2.unit_circle_or_not())
print(c5.unit_circle_or_not())

c1.translate(5, 5)
print(c1.x, c1.y)
c1.translate(0.5, 0.5)
print(c1.x, c1.y)

print(repr(c1))
print(str(c1))
```

#### Testing invalid circle

```python
try:
    c1 = Circle("5", 5, 5)
except TypeError as err:
    print(err)

try:
    c1 = Circle(5, "5", 5)
except TypeError as err:
    print(err)

try:
    c1 = Circle(5, 5, "5")
except TypeError as err:
    print(err)

try:
    c1 = Circle(True, 5, 5)
except TypeError as err:
    print(err)

try:
    c1 = Circle(-5)
except ValueError as err:
    print(err)

try:
    c1.area = 6
except AttributeError as err:
    print(err)
```

#### Testing circle comparisons

```python
c1 = Circle(5, 5, 5)
c2 = Circle(5)
c3 = Circle(6)

print(c1 == c2)
print(c1 == c3)
print(c1 < c3)
print(c1 > c3)
print(c1 <= c3)
print(c1 >= c3)
```
