- a function method can have different abilities based on their type

## Polymorphism class method example

```python
class Fish:
    def speak(self):
        print("blup, blup, blup")

    def __str__(self):
        return "fish swimming around"

class Fox:
    def speak(self):
        print("fox says")
    def __str__(self):
        return "Fox: no one knows how I sound"

fish = Fish()
print(Fish)
print(fish)
```

```python
animals = (Fox(), Fox(), Fish(), Fish())
for animal in animals:
    print(animal)
    animal.speak()
```

## Geometry

```python
import numpy as np

class Circle:
    def __init__(self, radius):
        self.radius = radius
    
    def area(self):
        return self.radius**2*np.pi
    
    def perimeter(self):
        return self.radius*2*np.pi

class Square:
    def __init__(self, side):
        self.side = side
    
    def area(self):
        return self.side**2

    def perimeter(self):
        return self.side*4

class Rectangle:
    def __init__(self, width, height):
        self.width = width
        self.height = height
    
    def area(self):
        return self.width*self.height
    
    def perimeter(self):
        return self.width*2+self.height*2
    
    def __repr__(self):
        return f"Width: {self.widht}, Height: {self.height}"
    

class Triangle:
    def __init__(self, base, height):
        self.base = base
        self.height = height
    
    def area(self):
        return (self.base*self.height)/2
    
    #def perimeter(self):



shapes = [Circle(1), Square(2), Rectangle(3,5), Rectangle(1,5)]
area = 0
perimeter = 0

for shape in shapes:
    area += shape.area()
    perimeter += shape.perimeter()

print(area)
print(perimeter)
```
