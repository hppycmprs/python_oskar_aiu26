## strings
# concatenate strings

```python
firstname = "bob"
lastname = "dylan"

print(firstname + " " + lastname)

print(f"{firstname} {lastname}")
```

```python
greet = f"""
Hello
{firstname.capitalize()} {lastname.capitalize()}
"""  # multiline string start """ end """

print(greet)
```

# string methods to convert string

```python
language = "   pYtHon   "

print(f"{language.capitalize() = }")
print(f"{language.upper() = }")
print(f"{language.lower() = }")
# method chaining
print(f"{language.strip().capitalize() = }")
```

# concatenate using loop

```python
bamba = ["pancake", "pokebowl", "fish n chips", "potatisbullar", "pytt i panna"]
days = ["mån", "tis", "ons", "tor", "fre", "sat"]

menu = "Veckomeny\n"
# concatenate strings to menu variable
for food, day in zip(bamba[1:], days[1:]):
    menu += f"{day}: {food}\n"

print(menu)
```

# Clean strings

```python
subjects = "   math,  BiOLOGy , PhYsics ,   programming    "

print(subjects)
print(subjects.strip())

subjects = subjects.strip().split(",")
print(subjects)

subjects = [subject.strip().upper() for subject in subjects]
print(subjects)
```

```python
for subject in subjects:
    print(f"{subject} is fun :)")
```
